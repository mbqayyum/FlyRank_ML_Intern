## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

I chose **Lane 2 — Refresh / Content Opportunity Scoring** because it bridges the gap between raw traffic declines and clear, operational actions for a content team. Instead of manually auditing thousands of legacy pages to guess which ones are slipping, we can use machine learning to build an automated prioritization queue. By pinpointing exactly which pages have high historical value but are losing ground due to content decay, we can maximize the return on editing resources and prevent long-term revenue loss.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**What decision does your work improve?**  
It improves the content team's investment decision: deciding which declining or underperforming pages are worth the time and money to rewrite, update, or optimize — rather than manually guessing or updating pages that have no potential for traffic recovery.

**Who acts on it?**  
Content strategists, SEO managers, and editorial teams who allocate writing resources.

**What does a wrong recommendation cost?**

- *False Positive* (high score given to a low-value or unrecoverable page): Wastes editorial hours and budget (~\$150–\$500 per article refresh) on a page that will never regain traffic.
- *False Negative* (low score given to a high-potential page): Results in missed opportunity costs — allowing high-value content to completely decay and leak revenue to competitors permanently.

**Why does data / ML help at all?**  
A simple rule like "refresh everything older than 180 days" ignores traffic momentum, engagement quality, keyword opportunity, and content type. The signals are many and interact — content age alone doesn't distinguish a page that is gracefully stable from one that is actively losing ground. ML earns its place because the pattern (which pages will benefit from a refresh) is real but too messy to capture with a single threshold.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
print(f"Dataset: {df.shape[0]:,} pages × {df.shape[1]} columns, {df['client_id'].nunique()} clients")
```

```python
# --- Number 1: How widespread is the decline problem? ---
n_total = len(df)
n_declining = (df["trend_direction"] == "down").sum()
pct_declining = n_declining / n_total * 100

print(f"1) {n_declining:,} of {n_total:,} pages ({pct_declining:.1f}%) are declining.")
print(f"   This affects {df.loc[df['trend_direction'] == 'down', 'client_id'].nunique()} "
      f"of {df['client_id'].nunique()} clients — the problem is nearly universal.")
```

```python
# --- Number 2: How much traffic is at risk? ---
declining = df["trend_direction"] == "down"
total_imp = df["impressions_90d"].sum()
declining_imp = df.loc[declining, "impressions_90d"].sum()

print(f"2) Declining pages account for {declining_imp:,.0f} of {total_imp:,.0f} "
      f"total 90-day impressions ({declining_imp / total_imp * 100:.1f}%).")
print(f"   That is ~{declining_imp / 1e6:.0f}M search impressions at risk of further decay.")
```

```python
# --- Number 3: The decline is steep, not marginal ---
median_trend = df.loc[declining, "trend_pct"].median()

print(f"3) The median declining page lost {abs(median_trend):.1f}% of its impressions "
      f"(last 30 days vs. prior 30 days).")
print(f"   These are not gentle drifts — a scoring model that catches them early")
print(f"   gives the content team a real head start.")
```

**Summary:** Over half the pages in the dataset are actively declining, they carry ~80M of ~156M total impressions, and their median drop is −55.6% month-over-month. A refresh-priority model that separates "worth saving" from "let it go" would directly reduce wasted editorial spend and protect revenue at scale.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What this work can claim (observed / directional / decision-support):**

- We can **observe** which measurable features (content age, days since last update, engagement rate, impression trends, keyword competition) are associated with pages that declined in the trailing window.
- We can build a **directional scoring model** that ranks pages by estimated refresh priority — pages scored higher are more likely to be experiencing meaningful traffic loss *and* to have recoverable opportunity.
- The output is a **decision-support tool**: a ranked queue that helps content teams allocate finite editing resources more efficiently than manual audits or blanket age-based rules.

**What this work cannot and will not claim:**

- **No causal proof.** Observing that older, less-engaged pages tend to decline does not prove that refreshing them *causes* recovery. A controlled experiment (refreshing a random subset and measuring lift) would be needed for that.
- **No "predicting Google."** We are not forecasting algorithmic changes or search-engine behavior. We score pages using observed, lagging signals — the model does not know what Google will do next.
- **No guarantees of ROI.** A high refresh-priority score means the page shows signs worth investigating; it does not promise that editing the page will recover traffic. External factors (competitor moves, intent shifts, seasonality) are outside the model's scope.
- **No client-level generalization (yet).** The model is trained on 32 pseudonymized clients. Performance on a new, very different client should be validated before acting on its scores.

**One-paragraph frame (per the skill template):**

> For **content strategists and SEO managers**, deciding **which declining pages to refresh first**, we will build a **priority-scoring model** from the **30k-page starter dataset** (and later the full warehouse), scoring pages by **observed decline severity and recoverable opportunity**, measured by **precision@K** (are the top-K recommended pages truly high-value decliners?). A wrong call costs **wasted editor hours** (false positive) or **missed recovery** (false negative). A plain rule isn't enough because **decline depends on the interaction of content age, engagement quality, keyword dynamics, and traffic momentum** — no single threshold captures it. We will claim only **observed, directional, decision-support** results.
